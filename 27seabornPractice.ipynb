```
import pandas as pd
import matplotlib.pyplot as plt
import seaborn as sns
```

```
sale = pd.read_csv('/content/sales (2).csv')
sale
```

Using sales, draw a lineplot of date vs sales.

```
sns.lineplot(data = sale, x = 'date', y = 'sales')
```

```
student = pd.read_csv('/content/students.csv')
student
```

From students, plot hours_studied vs marks and color by gender.

```
sns.scatterplot(data = student, x = 'hours_studied', y = 'marks', hue = 'gender')
```

Create a grid of scatterplots (hours_studied vs marks) faceted by subject.

```
sns.relplot(kind = 'scatter', data = student, x = 'hours_studied', y = 'marks', col = 'subject')
```

Plot a histogram of marks with a KDE overlay.

```
sns.histplot(data = student, x = 'marks', kde = True)
```

Show the KDE curve for hours_studied.

```
sns.kdeplot(data = student, x = 'hours_studied', hue = 'gender')
```

Display the count of records per subject.

```
sns.countplot(data = student, x = 'subject', hue = 'gender')
```

Draw a regression line for hours_studied and marks.

```
sns.regplot(data = student, x = 'hours_studied', y = 'marks')
```

```
sns.lmplot(data = student, x = 'hours_studied', y = 'marks', hue = 'gender')
```
