2D line plot

```
import numpy as np
import pandas as pd
import matplotlib.pyplot as plt
```

```
hours = [2,4,1,5,6,3,7]
days = [1,2,3,4,5,6,7]
```

```
plt.xlabel('Days')
plt.ylabel('Study hours')
plt.title('Study hours in a week')
plt.plot(days, hours)
```

```
hours_1 = [2,4,1,5,6,3,7]
hours_2 = [1,7,3,8,9,2,9]
days = [1,2,3,4,5,6,7]

plt.xlabel('Days')
plt.ylabel('Study hours')
plt.title('Study hours in a week')
plt.plot(days, hours_1)
plt.plot(days, hours_2)
```

2D line plot from pandas data

```
df = pd.read_csv('/content/enrollment_data.csv')
df
```

```
plt.plot(df['Year'], df['Programming'], label = 'Programming')
#show label using legend
plt.legend()
```

```
plt.plot(df['Year'], df['Programming'], label = 'Programming')
plt.plot(df['Year'], df['Digital Marketing'], label = 'DM')
#show label using legend
plt.legend(loc='upper right')
```

Adding some visualization

```
plt.plot(df['Year'], df['Programming'], label = 'Programming', color = "red", linewidth = 5, linestyle = 'dashed', marker = 'o', markersize = 7)
plt.plot(df['Year'], df['Digital Marketing'], label = 'DM')
#show label using legend
plt.legend(loc='upper right')
plt.grid()
plt.show()
```

Bar chart

```
data = {
    'Student':['A', 'B', 'C', 'D', 'E', 'F', 'G', 'H'],
    'Gender':['Male', 'Female', 'Male', 'Female', 'Male', 'Female', 'Male', 'Female'],
    'Study_hours':[1,2,3,4,5,6,7,8],
    'Attendance':[70, 60, 50, 45, 33, 44, 20, 10],
    'Grade':['Pass','Fail', 'Pass','Fail', 'Pass','Fail', 'Pass','Fail']
}
df = pd.DataFrame(data)
df
```

```
gender_group = df.groupby('Gender').size()
gender_group.index
gender_group.values
```

```
plt.bar(gender_group.index, gender_group.values, color=['red', 'orange'], edgecolor = 'black')
```

```
df = pd.read_csv('/content/student_data (1).csv')
df
```

```
status = df.groupby('CompletionStatus').size()
status
```

```
plt.bar(status.index, status.values, color=['red', 'orange'], edgecolor = 'black')
```

Pie chart

```
plt.pie(status, labels=status.index, autopct = '%1.1f%%', explode = (0.1, 0, 0), shadow = True)
plt.show()
```
