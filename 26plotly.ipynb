```
import plotly.express as px
import pandas as pd
```

```
student = pd.read_csv('/content/students.csv')
student
```

```
fig = px.scatter(student, x = 'hours_studied', y = 'marks', color = 'gender',size = 'previous_gpa', hover_data=['hours_studied', 'marks', 'subject'])
fig.show()
```

```
enrollment = pd.read_csv('/content/enrollment_data.csv')
enrollment
```

```
fig = px.line(enrollment, x = 'Year', y = 'Programming', markers = True)
fig.show()
```

```
fig.write_html('Programming_data.html')
```

```
fig = px.histogram(student, x = 'attendance_percentage', color = 'gender')
fig.show()
```

```
fig = px.histogram(student, x = 'attendance_percentage', color = 'gender', nbins = 5)
fig.show()
```
