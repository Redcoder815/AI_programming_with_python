```
import pandas as pd
```

```
df = pd.read_csv('/content/Practice.csv')
df
```

1. Add New Column and Save File

```
df['Discounted_price'] = df['Price'] * 0.9
df
```

```
df.to_csv('Discounted_Products.csv')
```

2. Check Duplicates and Remove Them

```
df.duplicated().sum()
```

```
clean_data = df.drop_duplicates()
print(clean_data)
```

3. Handle Missing Values

```
missing_count = df[['Age', 'Salary']].isnull().sum()
print(missing_count)
```

```
df['Age'] = df['Age'].fillna(df['Age'].mean())
df
```

```
drop_null = df.dropna(subset = 'Salary')
print(drop_null)
```

4. Apply Statistical Functions

```
df['Marks1'].min()
```

```
df['Total Marks'] = df.iloc[:,5:8].sum(axis=1)
df
```

```
df['Total Marks'].std()
```

```
statistic = df[['Marks1', 'Marks2', 'Marks3']].agg(['mean', 'max'])
print(statistic)
```
